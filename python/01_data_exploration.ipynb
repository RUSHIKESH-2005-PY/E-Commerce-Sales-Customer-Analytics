## 9. Outlier Analysis

### 9.1 Unit Price Outlier Analysis

```python
Q1 = df["Unit_Price"].quantile(0.25)
Q3 = df["Unit_Price"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
unit_price_outliers = df[
    (df["Unit_Price"] < lower_bound) |
    (df["Unit_Price"] > upper_bound)
]

print("Number of Unit Price Outliers:", len(unit_price_outliers))
```

Step 4.6 — Document the decision

### Delivery Days Cleaning Decision

The 13,256 missing `Delivery_Days` values are associated with orders having
`Cancelled`, `Shipped`, or `Processing` status.

These missing values are considered logically valid because delivery duration
may not be available for orders that were cancelled, are still processing, or
are still in transit.

Therefore, missing `Delivery_Days` values are retained as `NaN` rather than
being replaced with an estimated value.

```python
print(
    "Missing Delivery_Days after cleaning:",
    df_clean["Delivery_Days"].isnull().sum()
)
```

Step 4.7 — Coupon_Code

```python
coupon_missing = df_clean[df_clean["Coupon_Code"].isnull()]

print("Missing Coupon_Code records:", len(coupon_missing))
print("\nDiscount amount for missing coupon codes:")
print(coupon_missing["Discount_Amount"].describe())
```

```python
print("\nSample records:")
print(
    coupon_missing[
        ["Order_ID", "Coupon_Code", "Discount_Amount", "Net_Sales"]
    ].head(10)
)
```

### Coupon Code Cleaning Decision

There are 1,819 missing `Coupon_Code` values. Investigation shows that
these records can still have non-zero `Discount_Amount` values.

Therefore, missing `Coupon_Code` values cannot reliably be interpreted as
"No Coupon". The missing values are retained as `NaN` to avoid making an
unsupported assumption about the source of the discount.

```python
print(
    "Missing Coupon_Code after cleaning:",
    df_clean["Coupon_Code"].isnull().sum()
)
```

### Customer Rating Cleaning Decision

There are 1,453 missing `Customer_Rating` values. Existing ratings range
from 1.0 to 5.0, and missing ratings occur across multiple order statuses.

A missing rating cannot reliably be interpreted as any particular rating.
Therefore, the missing values are retained as `NaN` rather than being
replaced with the mean, median, or another assumed rating.

## Step 4.12 — Document the date decision

### Customer Signup Date Cleaning Decision

There are 9,349 orders where `Customer_Signup_Date` occurs after
`Order_Date`.

The dates are valid datetime values and neither date column contains
missing values. However, the relationship between the two dates violates
the expected business rule that a customer should normally sign up before
placing an order.

The correct signup dates cannot be determined from the available data.
Therefore, the original values are retained for traceability and are not
artificially modified or deleted.

```python
print("Invalid signup/order date records retained:",
      len(invalid_signup))
```

## 4.20 — Final Data Cleaning Summary
