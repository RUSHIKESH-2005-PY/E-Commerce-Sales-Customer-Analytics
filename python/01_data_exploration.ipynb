### 8.2 Net Sales Validation

```python
df["Calculated_Net_Sales"] = (
    df["Gross_Sales"] - df["Discount_Amount"]
)
```

```python
df["Net_Sales_Difference"] = (
    df["Net_Sales"] - df["Calculated_Net_Sales"]
)
```

```python
df["Net_Sales_Difference"].abs().describe()
```

### 8.3 Profit Validation

```python
df["Calculated_Profit"] = (
    df["Net_Sales"]
    - df["Cost_of_Goods_Sold"]
    - df["Shipping_Cost"]
)
```

```python
df["Profit_Difference"] = (
    df["Profit"] - df["Calculated_Profit"]
)
```

```python
df["Profit_Difference"].abs().describe()
```

## 9. Outlier Analysis

### 9.1 Unit Price Outlier Analysis

```python
Q1 = df["Unit_Price"].quantile(0.25)
Q3 = df["Unit_Price"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
unit_price_outliers = df[
    (df["Unit_Price"] < lower_bound) |
    (df["Unit_Price"] > upper_bound)
]

print("Number of Unit Price Outliers:", len(unit_price_outliers))
```

```python
unit_price_outliers[
    [
        "Order_ID",
        "Product_Category",
        "Product_Name",
        "Unit_Price",
        "Quantity",
        "Gross_Sales",
        "Net_Sales",
        "Profit"
    ]
].sort_values("Unit_Price", ascending=False).head(10)
```

```python
df.groupby("Product_Category")["Unit_Price"].agg(
    ["count", "min", "mean", "median", "max"]
).sort_values("max", ascending=False)
```

```python
df.groupby(
    ["Product_Category", "Product_Name"]
)["Unit_Price"].agg(
    ["count", "min", "median", "max"]
).sort_values("max", ascending=False).head(20)
```

## 13. Customer Signup-Date Validation

```python
invalid_signup_orders = df[
    df["Customer_Signup_Date"] > df["Order_Date"]
]

print("Orders before customer signup:", len(invalid_signup_orders))
```

```python
invalid_signup_orders[
    [
        "Order_ID",
        "Customer_ID",
        "Customer_Name",
        "Customer_Signup_Date",
        "Order_Date",
        "Product_Name",
        "Net_Sales"
    ]
].sort_values("Customer_Signup_Date", ascending=False).head(10)
```

```python
invalid_signup_orders["Customer_Signup_Date"].max()
```

```python
invalid_signup_orders["Customer_Signup_Date"].value_counts().head(10)
```

```python
invalid_signup_orders["Customer_ID"].nunique()
```

```python
customer_date_check = df.groupby("Customer_ID").agg(
    Signup_Date=("Customer_Signup_Date", "first"),
    First_Order_Date=("Order_Date", "min"),
    Order_Count=("Order_ID", "count")
)

customer_date_check["Signup_Before_First_Order"] = (
    customer_date_check["Signup_Date"] <=
    customer_date_check["First_Order_Date"]
)

customer_date_check.head(10)
```

```python
customer_date_check["Signup_Before_First_Order"].value_counts()
```

```python
customer_date_check["Signup_Order_Difference_Days"] = (
    customer_date_check["Signup_Date"]
    - customer_date_check["First_Order_Date"]
).dt.days

customer_date_check[
    customer_date_check["Signup_Before_First_Order"] == False
]["Signup_Order_Difference_Days"].describe()
```

```python
customer_date_check[
    customer_date_check["Signup_Before_First_Order"] == False
].sort_values(
    "Signup_Order_Difference_Days",
    ascending=False
).head(10)
```
